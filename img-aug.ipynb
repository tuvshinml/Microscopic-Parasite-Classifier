```python
import cv2
import matplotlib.pyplot as plt
import albumentations as A

img_path = "/content/ce1.jpg"

image = cv2.imread(img_path)
if image is None:
    raise ValueError("Image not found.")

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)


augmentations = [
    ("Original", None),
    ("Rotation", A.Rotate(limit=10, border_mode=cv2.BORDER_REFLECT, p=1.0)),
    ("Horizontal Flip", A.HorizontalFlip(p=1.0)),
    ("Vertical Flip", A.VerticalFlip(p=1.0)),
    ("Translation", A.Affine(
        translate_percent={"x": (-0.02, 0.02), "y": (-0.02, 0.02)},
        scale=1.0,
        rotate=0,
        border_mode=cv2.BORDER_REFLECT,
        p=1.0
    )),

    ("Scaling", A.Affine(
        scale=(0.97, 1.03),
        translate_percent=0,
        rotate=0,
        border_mode=cv2.BORDER_REFLECT,
        p=1.0
    )),

    ("Brightness", A.RandomBrightnessContrast(
        brightness_limit=0.08,
        contrast_limit=0,
        p=1.0
    )),

    ("Contrast", A.RandomBrightnessContrast(
        brightness_limit=0,
        contrast_limit=0.08,
        p=1.0
    )),

    ("Gaussian Noise", A.GaussNoise(
        std_range=(0.005, 0.015),
        mean_range=(0.0, 0.0),
        p=1.0
    )),

    ("Blur", A.GaussianBlur(blur_limit=(3, 3), p=1.0))
]


plt.figure(figsize=(18, 8))

for i, (title, aug) in enumerate(augmentations):

    if aug is None:
        aug_image = image.copy()
    else:
        aug_image = A.Compose([aug])(image=image)["image"]

    plt.subplot(2, 5, i + 1)
    plt.imshow(aug_image)
    plt.title(title, fontsize=10)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
